```python
file_path = "../data/raw/population_age_tainan_2025_12.csv"
```

```python
df = pd.read_csv(file_path)
```

```python
df.head()
```

```python
df.shape
```

```python
df["性別"].value_counts()
```

```python
df_total = df[df["性別"] == "計"].copy()
```

```python
df_total.shape
```

```python
df_district = df_total[df_total["區域別"] != "臺南市"].copy()
```

```python
df_district.shape
```

```python
columns_to_keep = [
    "區域別",
    "總計",
    "合計65至69歲",
    "合計70至74歲",
    "合計75至79歲",
    "合計80至84歲",
    "合計85至89歲",
    "合計90至94歲",
    "合計95至99歲",
    "100歲以上"
]
```

```python
columns_to_keep
```

```python
df_population = df_district[columns_to_keep].copy()
```

```python
df_population.shape
```

```python
df_population.head()
```

```python
df_population.dtypes
```

```python
age_65_plus_cols = [
    "合計65至69歲",
    "合計70至74歲",
    "合計75至79歲",
    "合計80至84歲",
    "合計85至89歲",
    "合計90至94歲",
    "合計95至99歲",
    "100歲以上"
]
```

```python
age_65_plus_cols
```

```python
df_population["population_65_plus"] = df_population[age_65_plus_cols].sum(axis=1)
```

```python
df_population[["區域別", "總計", "population_65_plus"]].head()
```

```python
(df_population["population_65_plus"] <= df_population["總計"]).all()
```

```python
df_population["pct_65_plus"] = (
    df_population["population_65_plus"]
    / df_population["總計"]
    * 100
)
```

```python
df_population[
    ["區域別", "總計", "population_65_plus", "pct_65_plus"]
].head()
```

```python
df_population[
    ["區域別", "總計", "population_65_plus", "pct_65_plus"]
]
```

```python
output_path = "../data/processed/population_tainan_2025_clean.csv"
```

```python
df_population.to_csv(output_path, index=False, encoding="utf-8-sig")
```
