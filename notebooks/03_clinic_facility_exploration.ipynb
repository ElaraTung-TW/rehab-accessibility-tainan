```python
lookup_path = "../data/raw/clinic_specialty_2024/欄位說明.csv"
```

```python
df_lookup = pd.read_csv(lookup_path)
```

```python
df_lookup.head()
```

```python
df_lookup.columns
```

```python
lookup_2024 = df_lookup[
    ["鄉鎮市區碼", "鄉鎮市區名稱(103年以後適用)"]
].copy()
```

```python
lookup_2024.columns = ["鄉鎮市區碼", "district_full"]
```

```python
lookup_tainan = lookup_2024[
    lookup_2024["鄉鎮市區碼"].between(501, 538)
].copy()
```

```python
lookup_tainan.shape
```

```python
lookup_tainan.head()
```

```python
df_clinic_tainan = tainan_check.merge(
    lookup_tainan,
    on="鄉鎮市區碼",
    how="left"
)
```

```python
df_clinic_tainan.shape
```

```python
df_rehab_clinic = df_clinic_tainan[
    ["district_full", "復健科"]
].copy()
```

```python
df_rehab_clinic.columns = [
    "district",
    "rehab_clinic_count"
]
```

```python
df_rehab_clinic["district"] = (
    df_rehab_clinic["district"]
    .str.removeprefix("臺南市")
)
```
