### Data Quality Note

The 2024 population CSV contains malformed columns named `歲人數` and
`歲人數以上`. Therefore, the 65+ population was not calculated by directly
summing the reported 65+ age columns.

Instead:

population aged 65+ = total population - population aged 0–64

The result was validated by comparing it with the sum of the 65–99 age
groups and checking the implied 100+ population.

```python
df_check = df[df["性別"] == "計"].copy()

df_check["age_sum_check"] = df_check[age_group_cols].sum(axis=1)

df_check["gap"] = (
    df_check["總計"]
    - df_check["age_sum_check"]
)

df_check[
    ["區域別", "總計", "age_sum_check", "gap", "歲人數", "歲人數以上"]
].head(10)
```

```python
under_65_cols = [
    "合計0至4歲人數",
    "合計5至9歲人數",
    "合計10至14歲人數",
    "合計15至19歲人數",
    "合計20至24歲人數",
    "合計25至29歲人數",
    "合計30至34歲人數",
    "合計35至39歲人數",
    "合計40至44歲人數",
    "合計45至49歲人數",
    "合計50至54歲人數",
    "合計55至59歲人數",
    "合計60至64歲人數"
]
```

```python
df_check = df[df["性別"] == "計"].copy()

df_check["population_under_65"] = (
    df_check[under_65_cols].sum(axis=1)
)

df_check["population_65_plus"] = (
    df_check["總計"]
    - df_check["population_under_65"]
)
```

```python
df_check[
    ["區域別", "總計", "population_under_65", "population_65_plus"]
].head(10)
```

```python
age_65_to_99_cols = [
    "合計65至69歲人數",
    "合計70至74歲人數",
    "合計75至79歲人數",
    "合計80至84歲人數",
    "合計85至89歲人數",
    "合計90至94歲人數",
    "合計95至99歲人數"
]

df_check["population_65_to_99"] = (
    df_check[age_65_to_99_cols].sum(axis=1)
)
```

```python
df_check["implied_100_plus"] = (
    df_check["population_65_plus"]
    - df_check["population_65_to_99"]
)
```

```python
df_check[
    [
        "區域別",
        "population_65_plus",
        "population_65_to_99",
        "implied_100_plus"
    ]
].head(10)
```

```python
district_check = df_check[
    df_check["區域別"] != "臺南市"
].copy()
```

```python
df_population_2024 = district_check[
    ["區域別", "總計", "population_65_plus"]
].copy()

df_population_2024.columns = [
    "district",
    "total_population",
    "population_65_plus"
]
```

```python
df_population_2024["pct_65_plus"] = (
    df_population_2024["population_65_plus"]
    / df_population_2024["total_population"]
    * 100
)
```
